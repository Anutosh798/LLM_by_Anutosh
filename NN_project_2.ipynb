```python
import numpy as np
from keras.models import Sequential
from keras.layers import Dense

X=np.array([[1],[2],[3],[4]])
y=np.array([2,4,6,8])

tiny_model=Sequential()
tiny_model.add(Dense(1,input_shape=(1,)))

tiny_model.compile(
    optimizer="adam",
    loss="mse"
)
tiny_model.fit(X,y,epochs=100)
```

```python
p=tiny_model.predict(np.array([[2]]))
print(p)
```

```python
weights=tiny_model.get_weights()
print(weights)
#so this model learned weight as 1.04 and biases as 0.09, our job is to make it like weights is 2 and biases 0 so lets change epochs
```

```python
tiny_model.fit(X,y,epochs=1000)
```

```python
weights,biases=tiny_model.get_weights()
print(weights)
print(biases)#weigths and biases are pretty close
```

### Let's go back to project

```python
np.random.seed(42)
tf.random.set_seed(42)
```

```python
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(30,activation="relu",input_shape=X_train.shape[1:]))

model.add(Dense(1))
model.summary()
```

```python
model.compile(
    loss="mean_squared_error",
    optimizer=keras.optimizers.SGD(learning_rate=1e-3)


)

history=model.fit(X_train,y_train,epochs=20,validation_data=(X_valid,y_valid))
```

```python
plt.plot(history.history['loss'],label="training loss")
plt.plot(history.history['val_loss'],label="validating loss")
plt.legend()
plt.xlabel("Epochs")
plt.ylabel("loss")
plt.title("training_validation_loss")
plt.show()
```

```python
model_test=model.evaluate(x_test,y_test)
```

```python
x_new=x_test[:5]
y_pred=model.predict(x_new)
print(y_pred[1])
print(y_pred[1][0])# for just number not []
for i in range(len(x_new)):
    print(f"sample {i+1}")
    print(f"Predicted {y_pred[i][0]}")
    print(f"actual {y_test[i]}")
```
